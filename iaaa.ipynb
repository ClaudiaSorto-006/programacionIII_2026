```python
#Librerias a import
import tensorflow as tf
import pandas as pd
import numpy as np
```

```python
#Leer el arcivo
datos = pd.read_csv('/content/conversor.csv', sep= ';', encoding='utf-8')
print(datos["celsius"])
```

```python
#Obtener los datos
c = datos["celsius"]
f = datos["farenheits"]
print(c, f)
```

```python
#crear el modelo
modelo = tf.keras.Sequential()
modelo.add(tf.keras.layers.Dense(units=1, input_shape=[1]))
```

```python
1 #compilar el modelo
2 modelo.compile(optimizer=tf.keras.optimizers.Adam(1.0), loss='mean_squared_error')
```

```python
# entrenar el modelo
epocas = modelo.fit(c, f, epochs=200, verbose=1)
```

```python
respuesta = modelo.predict(np.array([100.0]))
print(respuesta)
```
